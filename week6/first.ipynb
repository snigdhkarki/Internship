```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Loading Titanic dataset provided by Seaborn
df = sns.load_dataset('titanic')
df.head()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df['age'].dropna(), bins=20, color='skyblue', edgecolor='black')

plt.title('Distribution of Passenger Ages')
plt.xlabel('Age (Years)')
plt.ylabel('Frequency (Number of Passengers)')

plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['age'], df['fare'], alpha=0.5, color='coral')

plt.title('Passenger Fare vs. Age')
plt.xlabel('Age (Years)')
plt.ylabel('Fare Paid (£)')

plt.show()
```

```python
class_fare = df.groupby('pclass')['fare'].mean()

plt.figure(figsize=(8, 5))
plt.bar(class_fare.index.astype(str), class_fare.values, color=['gold', 'silver', '#cd7f32'])

plt.title('Average Ticket Fare by Passenger Class')
plt.xlabel('Passenger Class (Tier)')
plt.ylabel('Average Fare (£)')

plt.show()
```

```python
# Create age buckets for continuous trend
df['age_group'] = (df['age'] // 10) * 10
age_survival = df.groupby('age_group')['survived'].mean().sort_index()

plt.figure(figsize=(8, 5))
plt.plot(age_survival.index, age_survival.values, marker='o', linestyle='-', color='purple')

plt.title('Average Survival Rate by Age Group')
plt.xlabel('Age Group (Decades)')
plt.ylabel('Survival Rate (Probability 0.0 - 1.0)')

plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='pclass', y='age', data=df, palette='Set2')

plt.title('Age Distribution Across Passenger Classes')
plt.xlabel('Passenger Class (Tier)')
plt.ylabel('Age (Years)')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns
correlation_matrix = df[numeric_cols].corr()

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")

plt.title('Correlation Heatmap of Numerical Features')
plt.xlabel('Dataset Features (Variables)')
plt.ylabel('Dataset Features (Variables)')

plt.show()
```
